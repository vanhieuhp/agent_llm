```python
competitors = []
answers = []
messages = [{"role": "user", "content": content}]
```

```python
def record(model_name, answer):
    competitors.append(model_name)
    answers.append(answer)
    display(Markdown(answer))
```

```python
gpt_model_name = os.getenv("OPENAI_MODEL")
print(f"Using {gpt_model_name} model")
response = chatgpt.chat.completions.create(
    model=gpt_model_name,
    messages=messages
)

answer = response.choices[0].message.content
record(gpt_model_name, answer)
```

```python
deepseek_model = os.getenv("DEEPSEEK_MODEL")
response = deepseek.chat.completions.create(
    model=deepseek_model,
    messages=messages
)
answer = response.choices[0].message.content
record(deepseek_model, answer)
```

```python
token_harbor_model = os.getenv("TOKEN_HARBOR_MODEL")
response = token_harbor.chat.completions.create(
    model=token_harbor_model,
    messages=messages
)
answer = response.choices[0].message.content
record(token_harbor_model, answer)
```
